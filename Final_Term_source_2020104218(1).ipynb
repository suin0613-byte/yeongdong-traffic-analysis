```python
# 구간별 비교
import pandas as pd

df = pd.read_excel('10월_상행_가공\\10월_상행_가공.xlsx')

# 구간 일치시켜주기
df = df.drop(df[df['구간'].isin(['군자JC ~ 월곶JC', '북수원IC ~ 부곡IC', '동군포IC ~ 군포IC', '군포IC ~ 둔대JC', '안산JC ~ 안산IC', '안산IC ~ 서안산IC'])].index)
# 교통량 원본 데이터에 23일과 24일에 대한 데이터가 삭제되어 있어서 똑같이 행을 맞춰줌
df = df.drop(df[df['10월'].isin([23, 24])].index)

df.to_excel('10월_교통량_가공\\평균속도_가공_상행.xlsx', index=False)

df['평균속도'] = df.mean(axis=1, numeric_only= True)

# 구간별 평균속도 구하기
section_data = []
valid_days = [day for day in range(1, 32) if day not in [23, 24]]
for day in valid_days:
    day_section = df.loc[df['10월'] == day]
    for section in day_section['구간'].unique():
        section_data.append({
            '구간': section,
            '평균속도': day_section.loc[day_section['구간'] == section, '평균속도'].mean()
        })

original_sections_speed = df['구간'].unique().tolist()  
average_speeds = df.groupby('구간')['평균속도'].mean().astype(int).reset_index()

# 원래 구간 순서대로 정렬
average_speeds['구간'] = pd.Categorical(average_speeds['구간'], categories=original_sections_speed, ordered=True)
average_speeds = average_speeds.sort_values('구간').reset_index(drop=True)

# 구간별 교통량 구하기
df_traffic = pd.read_excel('10월_교통량_가공\\교통량_가공_상행.xlsx')

df_traffic['교통량'] = df_traffic.mean(axis=1, numeric_only= True).astype(int)

section_data = []
valid_days = [day for day in range(1, 32) if day not in [23, 24]]
for day in valid_days:
    day_section = df_traffic.loc[df_traffic['10월'] == day]
    for section in day_section['구간명'].unique():
        section_data.append({
            '구간명': section,
            '교통량': day_section.loc[day_section['구간명'] == section, '교통량'].mean().astype(int)
        })

# 구간별 평균속도 데이터 프레임과 구간별 교통량 데이터 프레임 형식 맞춰주기
df_traffic = df_traffic[~df_traffic['구간명'].str.contains('마성IC-신갈JC')]
original_sections_Traffic = df_traffic['구간명'].unique().tolist()  
average_Traffic = df_traffic.groupby('구간명')['교통량'].mean().astype(int).reset_index()

average_Traffic['구간명'] = pd.Categorical(average_Traffic['구간명'], categories=original_sections_Traffic, ordered=True)
average_Traffic = average_Traffic.sort_values('구간명').reset_index(drop=True)
average_Traffic.rename(columns= {'구간명' : '구간'}, inplace= True)

section_dict = {
    '신갈JC ~ 동수원IC' : '신갈JC-동수원IC',
    '동수원IC ~ 북수원IC' : '동수원IC-북수원IC',
    '부곡IC ~ 동군포IC' : '부곡IC-동군포IC',
    '둔대JC ~ 안산JC' : '둔대JC-안산JC',
    '서안산IC ~ 군자T/G' : '서안산IC-군자TG',
    '군자T/G ~ 군자JC' : '군자TG-군자JC',
    '월곶JC ~ 서창JC' : '월곶JC-서창JC'
}
average_speeds['구간'] = average_speeds['구간'].replace(section_dict)
Converged_data = pd.merge(average_Traffic, average_speeds, on='구간')
Converged_data.to_excel('10월_교통량_가공\\구간별_교통량_속도.xlsx')

Converged_data
```
